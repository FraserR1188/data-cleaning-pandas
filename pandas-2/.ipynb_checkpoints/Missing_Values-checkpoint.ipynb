```python
import pandas as pd
```

```python
data_df = pd.read_csv("hotel_bookings.csv")
```

```python
data_df = pd.DataFrame(data_df)
```

```python
data_df.isnull().sum()
```

To find the percentage of total missing values, carry out the equation below:

```python
(data_df.isnull().sum()/len(data_df))*100
```

```python
# removing the company feature
data_df.drop("company", inplace=True, axis=1)
```

```python
data_df
```

```python
pd.set_option("display.max_columns", None)
```

```python
data_df
```

```python
# unique values in children feature
data_df["children"].value_counts()
```

```python
# filling children feature with 0 - lesson mentioned this is the different way to call the column.
data_df.children = data_df.children.fillna(0)
```

```python
data_df
```

```python
data_df["country"].value_counts()
```

```python
# dropping all rows with missing values in the country feature
data_df = data_df.dropna(subset=["country"])
```

```python
data_df
```

```python
data_df = data_df.reset_index(drop=True)
```

```python
data_df
```
