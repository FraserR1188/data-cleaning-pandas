# Dealing with cetgorical data

```python
# creating a dataframe with just the categorical features from our original dataframe

import pandas as pd
```

```python
data_df = pd.read_csv("hotel_bookings.csv")
```

```python
data_df = pd.DataFrame(data_df)
```

```python
data_df.head()
```

```python
cat_cols = [col for col in data_df.columns if data_df[col].dtype=="object"]
cat_cols
```

```python
cat_df = data_df[cat_cols]
cat_df
```

```python
# printing unique values of each column
for col in cat_df.columns:
    print(f"{col}: \n {cat_df[col].unique()}\n")
```

```python
cat_df["arrival_date_month"] = cat_df["arrival_date_month"].map({"January":1, "February":2, "March":3, "April":4, "May":5, "June":6, "July":7, "August":8, "September":9, "October":10, "November":11, "December":12})
```

```python
cat_df.head()
```

```python
# country and hotel features

from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()

# label encoding country

cat_df["country"] = le.fit_transform(cat_df["country"])

# label encoding hotel

cat_df["hotel"] = le.fit_transform(cat_df["hotel"])
```

```python
cat_df
```

```python
# one-hot encoding
cat_df = pd.get_dummies(data=cat_df, columns=["meal", "market_segment", "distribution_channel", "reserved_room_type", "assigned_room_type", "customer_type", "deposit_type", "reservation_status"])
```

```python
cat_df
```

```python
# numerical dataframe
num_df = data_df.drop(columns=cat_cols, axis=1)
```

```python
num_df
```

```python
final_df = pd.concat([num_df, cat_df], axis=1)
```

```python
# Correlation Analysis

# create a correlation matrix

corr_matrix = final_df.corr().abs()
corr_matrix
```
