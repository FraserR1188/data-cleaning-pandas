```python
houses = pd.read_csv("homes.csv")
houses.head()
```

```python
houses.plot()
```

```python
houses.plot(x='List', y='Rooms', marker='o', markerfacecolor='red')
```

```python
# Scatter Plot

houses.plot(x='List', y='Rooms', kind='scatter')
```

```python
# List houses which are less than 50 years of Age

houses[houses['Age']<=50].plot(kind='scatter', x='Age', y='List')
```

```python
houseplotdata = houses[houses['Age']<=50]
houseplotdata = houseplotdata[houseplotdata['List']<=300]
houseplotdata.plot(kind='scatter', x='Age', y='List')
```

```python
enrollment = pd.read_csv('enrollment.csv')
enrollment.head()
```

```python
enrollment.plot(kind="area", x="Year")
```

```python
pivotenrolled = enrollment.pivot(index='Year', columns='Class', values='Enrollment')
pivotenrolled.head()
```

```python
pivotenrolled.plot(kind="area")
```

```python
pivotenrolled.plot(kind="area", stacked=False)
```
